```python
# df['Review Rating']=df.groupby ('Category')['Review Rating'].transform(lambda x: x.fillna(x.median))

df["Review_Rating"] = pd.to_numeric(
    df["Review_Rating"],
    errors="coerce"
)

median_value = df["Review_Rating"].median()

df["Review_Rating"] = df["Review_Rating"].fillna(median_value)
```

```python
df.isnull().sum()
```

```python
# snake casing

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace( ' ','_')
df=df.rename(columns ={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# feature engineering 
# creating a column age_group

labels =['Young Adult' , 'Adult' , 'Middle-age', 'Senior']
df['age_groups']=pd.qcut(df['age'],q=4  , labels=labels)
```

```python
df[['age','age_groups']].head()
```

```python
#create column purchase_frequency_days 
frequency_mapping ={
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' :30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' :365,
    'Every 3 Months' :90
}
df['purchase_frequancy_days'] = df['frequency_of_purchases'].map(frequency_mapping)
    
```
